Ec = (1/2) * m * (v^2)

```python
# Melhorado
import numpy as np

# produz mesma sequência de números aleatórios no código |  N é o número de amostras
np.random.seed(42)
N = 100

# Números aleatórios com distribuição uniforme (inicio, fim, amostras)
m = np.random.uniform(1, 10, N) 
v = np.random.uniform(1, 30, N)

Ec = 0.5 * m * (v**2)

print(Ec)
```

```python
import matplotlib.pyplot as plt

fig,ax = plt.subplots(2)

ax[0].scatter(m, Ec, color='b')
ax[0].set_title('Energia Cinética x Massa')
ax[0].set_xlabel('Massa (kg)')
ax[0].set_ylabel('Energia Cinética (J)')
ax[0].grid(True)    # ativa a grade

ax[1].scatter(v, Ec, color='g')
ax[1].set_title('Energia Cinética x Velocidade')
ax[1].set_xlabel('Velocidade (m/s)')
ax[1].set_ylabel('Energia Cinética (J)')
ax[1].grid(True)    # ativa a grade

# Um gráfico ficou em cima do outro 
# Arrumo a sobreposição
plt.tight_layout()
plt.show()
```
